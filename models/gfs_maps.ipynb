```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path("../src")))

from gfs import GfsFetcher, GribReader, MapBuilder
from gfs.products import MLCape, ThetaEProduct, ConvectionProduct, DailyConvectionProduct, MslpProduct, ConvectivePrecipProduct
```

```python
DATA_DIR = Path("../data/gfs")
N_FILES = 24*3
```

```python
fetcher = GfsFetcher(output_dir=DATA_DIR)
file_paths = fetcher.fetch_latest(n_files=N_FILES, clear_first=True)
print(f"Pobrano {len(file_paths)} plików GRIB.")
```

```python
# Sprawdz dostepne parametry w pierwszym pliku GRIB
from gfs.reader import GribReader

first_file = sorted(file_paths)[0]
df = GribReader(first_file).scan_file()
print(f"Plik: {first_file.name}")
df.sort_values(["typeOfLevel", "parameter"]).reset_index(drop=True)
```

```python
# Generuj mapy dla wszystkich krokow prognozy
product = ConvectivePrecipProduct()

for fp in sorted(file_paths):
    product.plot(fp, save=True, show=False)

print(f"Wygenerowano {len(file_paths)} map.")
```
